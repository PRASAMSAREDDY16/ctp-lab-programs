```python
from dataclasses import dataclass

@dataclass
class Stack:
    a: list

    def push(self, x): self.a.append(x)
    def pop(self): return self.a.pop()

@dataclass
class Queue:
    a: list

    def enqueue(self, x): self.a.append(x)
    def dequeue(self): return self.a.pop(0)

s = Stack([])
s.push(10); s.push(20)
print(s.a, s.pop())

q = Queue([])
q.enqueue(10); q.enqueue(20)
print(q.a, q.dequeue())
```

```python
def maximum(a: int, b: int) -> int:
    if a > b:
        return a
    return b


def test_maximum():
    assert maximum(10, 20) == 20


def test_equal():
    assert maximum(5, 5) == 5


# Run tests
test_maximum()
test_equal()

print("Maximum of 10 and 20:", maximum(10, 20))
print("All tests passed!")
```

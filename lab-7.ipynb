```python
import threading
from queue import Queue

q = Queue()
lock = threading.Lock()

def producer():
    for i in range(5):
        with lock:
            q.put(i)
            print("Produced:", i)

def consumer():
    for i in range(5):
        with lock:
            print("Consumed:", q.get())

t1 = threading.Thread(target=producer)
t2 = threading.Thread(target=consumer)

t1.start()
t2.start()
t1.join()
t2.join()
```
